```python
import sqlite3
import pandas as pd

db_name = "/data/finlm_sequences/finlm_chunks_450.sqlite"
conn = sqlite3.connect(db_name)
table_names = pd.read_sql("Select name from sqlite_master Where type = 'table'", conn)
conn.close()
table_names
```

```python
for t_name in table_names.name.tolist():
    conn = sqlite3.connect(db_name)
    n_seqs = pd.read_sql(f"Select MAX(ROWID) From {t_name}", conn)
    examples = pd.read_sql(f"Select * From {t_name} Limit 5", conn)
    conn.close()
    print(t_name)
    print("="*100)
    print(f"Number of sequences: {n_seqs.values[0][0]}")
    for example in examples.sequence:
        print(example)
    print("="*100)
```
